```python
import pandas as pd

# 1. Carregar os dados
# Arquivo CSV com as predições do modelo
df_pred = pd.read_csv('test1modelo2.csv')

# Arquivo Excel com os rótulos reais (gabarito)
df_real = pd.read_excel('test1-rotulado-para-teste-estatistico.xlsx')

# 2. Renomear as colunas de classificação para evitar confusão na hora de juntar
df_pred = df_pred.rename(columns={'clarity': 'classe_predita'})
df_real = df_real.rename(columns={'clarity': 'classe_real'})

# 3. Unir os dataframes usando o texto da resposta como chave (ou index, se estiverem na mesma ordem)
# Aqui faremos o merge pelo texto para garantir o alinhamento correto
df_comparacao = pd.merge(df_real, df_pred, on='resp_text', how='inner')

# 4. Filtrar os casos de classificação errônea (onde a predição é diferente do rótulo real)
df_erros = df_comparacao[df_comparacao['classe_real'] != df_comparacao['classe_predita']]

# 5. Visualizar os primeiros exemplos de erros
print(f"Total de classificações errôneas encontradas: {len(df_erros)}")
display(df_erros.head())

# Opcional: Exportar os erros para um novo CSV para facilitar a análise qualitativa
df_erros.to_csv('erros_classificacao_analise.csv', index=False, encoding='utf-8')
print("\nArquivo 'erros_classificacao_analise.csv' salvo com sucesso!")
```
