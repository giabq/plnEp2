```python
import pandas as pd
import re

def limpar_texto_basico(texto):
    if not isinstance(texto, str):
        return texto
    texto = texto.lower()
    texto = re.sub(r'http\S+|www\.\S+', '', texto)
    texto = re.sub(r'[@#]\w+', '', texto)
    texto = re.sub(r'[^\w\s]', ' ', texto)
    texto = re.sub(r'\s+', ' ', texto).strip()
    return texto

def processar_dataset(caminho_entrada, caminho_saida, coluna_texto='resp_text'):
    print(f"Processando o arquivo: {caminho_entrada}...")

    if caminho_entrada.endswith('.xlsx'):
        df = pd.read_excel(caminho_entrada)
    else:
        df = pd.read_csv(caminho_entrada)

    df[coluna_texto] = df[coluna_texto].apply(limpar_texto_basico)

    if 'clarity' not in df.columns:
        df['clarity'] = None

    colunas_para_manter = ['resp_text', 'clarity']
    colunas_presentes = [col for col in colunas_para_manter if col in df.columns]
    df = df[colunas_presentes]

    df.to_csv(caminho_saida, index=False)
    return df

df_train = processar_dataset('/content/train.csv', '/content/train_limpo.csv')
print("Amostra Treino:")
display(df_train.head())
```

```python
df_test1 = processar_dataset('/content/test1-rotulado-para-teste-estatistico.xlsx', '/content/test1_limpo.csv')
print("Amostra Teste 1:")
display(df_test1.head())

df_test2 = processar_dataset('/content/test2.xlsx', '/content/test2_limpo.csv')
print("\nAmostra Teste 2 (sem rótulos preenchidos):")
display(df_test2.head())
```
